```python
# Linear Regression - Supervised
```

```python
#equation --> y=mx+c (straight-line)
#y--> output
#X--> Input
#m--> slope
#c--> intercept
```

```python
#x- Independent Y- Dependent
```

```python
#STEP - IMPORT
from sklearn.linear_model import LinearRegression
import numpy as np
```

```python
#SIMPLE EASY
#dataset
X=np.array([[1000],[1200],[1500],[1800],[2000]])
#target/output
y=np.array([10,12,15,18,20])
```

```python
#create model
model=LinearRegression()
```

```python
#Train model
model.fit(X,y)
```

```python
#prediction/find
prediction=model.predict([[1700]])
```

```python
#print
print("predicted price:",prediction[0],"Lakhs")
```

```python
print("slope:",model.coef_) #m
```

```python
print("Intercept:",model.intercept_) #c
```

```python
import pandas as pd
```

```python
from sklearn.linear_model import LinearRegression
```

```python
#dataset using dataframe
data={
    "Area":[1000,1200,1500,1800,2000],
    "Bedrooms":[2,2,3,3,4],
    "Price":[20,24,30,36,40]
}
```

```python
df=pd.DataFrame(data)
```

```python
#X- independent variable - area, bedroom
#y- dependent variable- price
```

```python
#features
X2=df[["Area","Bedrooms"]]
y2=df["Price"]
```

```python
#create model
model2=LinearRegression()

#train
model2.fit(X2,y2)
```

```python
#predict
prediction2=model2.predict([[1950,3]])
```

```python
#print
print("predicted Price:", prediction2[0],"lakhs")
```
